```python
sns.countplot(data=df, x = "Churn Label")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")

plt.show()
```

```python
#How does churn vary across different customer characteristics?
pd.crosstab(
    df["Senior Citizen"],
    df["Churn Label"],
    normalize="index") * 100
```

```python
sns.countplot(
    data=df,
    x="Senior Citizen",
    hue="Churn Label"
)

plt.title("Churn by Senior Citizen Status")
plt.show()
```

```python
#How is customer tenure distributed?
#Do churned customers have different tenure patterns?
sns.histplot(
    data=df,
    x="Tenure Months",
    bins=30
)

plt.title("Customer Tenure Distribution")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Churn Label",
    y="Tenure Months"
)

plt.title("Tenure by Churn Status")
plt.show()
```

```python
#How does customer churn vary across contract types?
sns.countplot(
    data=df,
    x="Contract",
    hue="Churn Label"
)

plt.title("Churn Distribution by Contract Type")
plt.xticks(rotation=15)
plt.show()
```

```python
contract_churn = (
    df.groupby("Contract")["Churn Value"]
      .mean() * 100
)

contract_churn
```

```python
#Charges & Customer Value
sns.boxplot(
    data=df,
    x="Churn Label",
    y="Monthly Charges"
)

plt.title("Monthly Charges by Churn Status")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Churn Label",
    y="CLTV"
)

plt.title("CLTV by Churn Status")
plt.show()
```

```python
#Churn Reasons
churned = df[df["Churn Label"] == "Yes"]
reason_counts = (
    churned["Churn Reason"]
    .value_counts()
    .head(10)
)
```

```python
reason_counts.sort_values().plot(kind="barh")

plt.title("Top 10 Churn Reasons")
plt.xlabel("Number of Customers")
plt.ylabel("Churn Reason")
plt.show()
```

```python
#Numerical Relationships
numeric_cols = [
    "Tenure Months",
    "Monthly Charges",
    "Total Charges",
    "CLTV",
    "Churn Score"
]
```

```python
correlation = df[numeric_cols].corr()

correlation
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")
plt.show()
```

```python
df[numeric_cols].dtypes
```

```python
df[numeric_cols].isnull().sum()
```

```python
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```
